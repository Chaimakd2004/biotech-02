```python
import pandas as pd
import os

# Chemin du fichier CSV sur votre Google Drive
chemin_csv = "/content/drive/MyDrive/offlineDB2_wordConvNeXt.csv"

# Vérification et lecture du fichier
if os.path.exists(chemin_csv):
    df = pd.read_csv(chemin_csv)
    print("Fichier CSV chargé avec succès ! Voici les premières lignes :")
    display(df.head())
else:
    print("Le fichier CSV est introuvable au chemin indiqué. Veuillez vérifier que votre Google Drive est bien monté.")
```

```python
# 1. Analyse rapide des colonnes du jeu de données
print(f"Nombre total de lignes : {df.shape[0]}")
print(f"Nombre total de colonnes : {df.shape[1]}")

# Affichage des types des dernières colonnes pour repérer la cible (Y)
print("\nTypes des 5 dernières colonnes du DataFrame :")
display(df.iloc[:, -5:].dtypes)

# Affichage des valeurs uniques de la toute dernière colonne pour voir s'il s'agit des classes/labels
derniere_colonne_nom = df.columns[-1]
print(f"\nExemples de valeurs dans la dernière colonne ({derniere_colonne_nom}) :")
print(df[derniere_colonne_nom].head())
print(f"\nNombre de valeurs uniques dans '{derniere_colonne_nom}' : {df[derniere_colonne_nom].nunique()}")
```

```python
# 2. Séparation automatique de X (caractéristiques) et Y (cible)
# On suppose ici que la dernière colonne est notre variable cible (Y)
Y = df.iloc[:, -1]          # Sélection de la dernière colonne
X = df.iloc[:, :-1]         # Sélection de toutes les colonnes sauf la dernière

print("Séparation effectuée avec succès !")
print(f"Dimensions de X (Features) : {X.shape}")
print(f"Dimensions de Y (Target)   : {Y.shape}")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA

# 1. Extraction des informations cliniques/biologiques de Y (chemin d'image)
# On extrait l'identifiant de l'utilisateur (ex: u001, u002) qui correspond à la signature biologique du scripteur
subjects = Y.apply(lambda path: path.split('/')[-1].split('_')[0])

print("--- ANALYSE CLINIQUE ET BIOLOGIQUE ---")
print(f"Nombre de sujets uniques analysés (patients/scripteurs) : {subjects.nunique()}")
print("Répartition des échantillons par sujet (Top 10) :\n", subjects.value_counts().head(10))
```

```python
# 2. Réduction de dimension (PCA) pour visualiser l'espace des caractéristiques d'écriture (X)
# On réduit les 769 dimensions de ConvNeXt en 2 composantes principales pour les cartographier
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

# Création d'un DataFrame de visualisation
df_visu = pd.DataFrame(X_pca, columns=['Composante Motrice 1', 'Composante Motrice 2'])
df_visu['Sujet'] = subjects.values

# Affichage de la carte de distribution des caractéristiques de l'écriture
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df_visu, x='Composante Motrice 1', y='Composante Motrice 2', hue='Sujet', palette='viridis', legend=False, alpha=0.7)
plt.title("Visualisation de l'Empreinte Biomotrice de l'Écriture (PCA)")
plt.xlabel("Variabilité globale de la forme/pression (PC1)")
plt.ylabel("Variabilité des micro-mouvements (PC2)")
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

print(f"Explication : Les deux premiers axes de la PCA capturent {pca.explained_variance_ratio_.sum()*100:.2f}% de la variance totale des mouvements d'écriture.")
```

```python
# 1. Analyse textuelle et sémantique de Y (Nature de la variable cible)
print("--- ANALYSE DE LA NATURE DE Y ---")
print(f"Type de données de Y : {Y.dtype}")

# Extraction des dossiers parents et des structures de noms
paths_split = Y.apply(lambda p: p.split('/'))
df_nature = pd.DataFrame({
    'chemin_complet': Y,
    'nom_fichier': Y.apply(lambda p: p.split('/')[-1]),
    'sujet': subjects # déjà extrait précédemment
})

# Détection des classes (Sain 'H' pour Healthy vs Dysgraphique 'D' ou similaire si encodé dans le nom)
# Analysons si des étiquettes cliniques communes se trouvent dans le nom du fichier
def detecter_classe(nom):
    nom_lower = nom.lower()
    if 'sain' in nom_lower or '_h_' in nom_lower or 'healthy' in nom_lower:
        return 'Sain (Healthy)'
    elif 'dys' in nom_lower or '_d_' in nom_lower or 'patient' in nom_lower:
        return 'Pathologique (Dysgraphie)'
    else:
        return 'Non spécifié (À classifier)'

df_nature['Statut_Estime'] = df_nature['nom_fichier'].apply(detecter_classe)

print("\nExemples de décomposition des chemins de Y :")
display(df_nature.head())

print("\nRépartition de la nature détectée :")
display(df_nature['Statut_Estime'].value_counts())
```

```python
# 2. Visualisation graphique de la répartition de Y
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))
sns.countplot(data=df_nature, x='Statut_Estime', palette='Set2')
plt.title("Nature et distribution des classes détectées dans Y")
plt.xlabel("Catégorie / Statut")
plt.ylabel("Nombre d'échantillons")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score
from sklearn.model_selection import train_test_split

# 1. Utilisation de la vraie colonne cible 'label' présente dans le dataset
Y_true = df['label']
X_features = df.drop(columns=['label', 'image_path'])

# 2. Nouvelle séparation 80% entraînement / 20% test avec stratification
X_train, X_test, Y_train_labels, Y_test_labels = train_test_split(
    X_features,
    Y_true,
    test_size=0.20,
    random_state=42,
    stratify=Y_true
)

print("--- ENTRAÎNEMENT DES MODÈLES AVEC LA COLONNE 'LABEL' ---")
print(f"Distribution des classes dans l'entraînement :\n{Y_train_labels.value_counts()}")

# 3. Entraînement du modèle Random Forest
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, Y_train_labels)
rf_preds = rf_model.predict(X_test)

# 4. Entraînement du modèle Régression Logistique
lr_model = LogisticRegression(max_iter=1000, random_state=42)
lr_model.fit(X_train, Y_train_labels)
lr_preds = lr_model.predict(X_test)

# --- ÉVALUATION DES PERFORMANCES ---
print("\n=== RÉSULTATS - RANDOM FOREST ===")
print(f"Précision Globale (Accuracy) : {accuracy_score(Y_test_labels, rf_preds)*100:.2f}%")
print(classification_report(Y_test_labels, rf_preds))

print("\n=== RÉSULTATS - RÉGRESSION LOGISTIQUE ===")
print(f"Précision Globale (Accuracy) : {accuracy_score(Y_test_labels, lr_preds)*100:.2f}%")
print(classification_report(Y_test_labels, lr_preds))
```

```python
from sklearn.model_selection import cross_val_score
from sklearn.metrics import confusion_matrix, roc_auc_score, RocCurveDisplay
import matplotlib.pyplot as plt
import numpy as np

# 1. Validation croisée sur l'ensemble complet pour vérifier la stabilité (5-fold cross-validation)
print("--- VALIDATION STATISTIQUE : CROSS-VALIDATION (5 Plis) ---")

rf_cv_scores = cross_val_score(rf_model, X_features, Y_true, cv=5)
lr_cv_scores = cross_val_score(lr_model, X_features, Y_true, cv=5)

print(f"Précisions Random Forest par pli : {rf_cv_scores}")
print(f"Précision moyenne Random Forest : {rf_cv_scores.mean()*100:.2f}% (+/- {rf_cv_scores.std()*100:.2f}%)")

print(f"\nPrécisions Régression Logistique par pli : {lr_cv_scores}")
print(f"Précision moyenne Régression Logistique : {lr_cv_scores.mean()*100:.2f}% (+/- {lr_cv_scores.std()*100:.2f}%)")

# 2. Calcul des matrices de confusion sur l'ensemble de test
print("\n--- MATRICES DE CONFUSION ---")
rf_cm = confusion_matrix(Y_test_labels, rf_preds)
lr_cm = confusion_matrix(Y_test_labels, lr_preds)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Affichage de la matrice de confusion de Random Forest
sns.heatmap(rf_cm, annot=True, fmt='d', cmap='Blues', ax=axes[0])
axes[0].set_title("Matrice de Confusion : Random Forest")
axes[0].set_xlabel("Prédictions")
axes[0].set_ylabel("Vraies Valeurs")

# Affichage de la matrice de confusion de la Régression Logistique
sns.heatmap(lr_cm, annot=True, fmt='d', cmap='Greens', ax=axes[1])
axes[1].set_title("Matrice de Confusion : Régression Logistique")
axes[1].set_xlabel("Prédictions")
axes[1].set_ylabel("Vraies Valeurs")

plt.tight_layout()
plt.show()

# 3. Calcul du score de performance ROC-AUC
rf_probs = rf_model.predict_proba(X_test)[:, 1]
lr_probs = lr_model.predict_proba(X_test)[:, 1]

rf_auc = roc_auc_score(Y_test_labels, rf_probs)
lr_auc = roc_auc_score(Y_test_labels, lr_probs)

print(f"\nScore ROC-AUC (Random Forest) : {rf_auc:.4f}")
print(f"Score ROC-AUC (Régression Logistique) : {lr_auc:.4f}")
```
